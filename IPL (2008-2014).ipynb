```python
import pandas as pd
import numpy as np

df=pd.read_csv(r'C:\Users\X1 Carbon\OneDrive\Desktop\matches.csv')
print(df.shape)
```

```python
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
df['date']=pd.to_datetime(df['date'],format='%Y-%m-%d')
```

```python

print(df['date'].dtype)
```

```python
#new column
df['year']=df['date'].dt.year
```

```python
# Remove single column

df = df.drop('season', axis=1)
```

```python
df['city'] = df['city'].fillna('Unknown')
```

```python
df['player_of_match'] = df['player_of_match'].fillna('No Award')
```

```python
df['winner'] = df['winner'].fillna('No Result')
```

```python
print(df['result_margin'].unique())
print(df['target_runs'].unique())
print(df['target_overs'].unique())
```

```python
df[df['result_margin'].isna() | df['target_runs'].isna() | df['target_overs'].isna()][['date','team1','team2','result', 'winner', 'result_margin', 'target_runs', 'target_overs']]
```

```python
df['result_margin'] = df['result_margin'].fillna(0)
df['target_runs']   = df['target_runs'].fillna(0)
df['target_overs']  = df['target_overs'].fillna(0)

# here i filled the NaN values of columns [result_margin, target_runs, target_overs] to 0 means 
# those matches might be cancelled or they are tie.
```

```python
df[df['result_margin'].isna() | df['target_runs'].isna() | df['target_overs'].isna()]
```

```python
df['method'] = df['method'].replace('NA', np.nan)
df['method'] = df['method'].fillna('Normal')
```

```python
print(df.isnull().sum())
```

```python
print(df.dtypes)
```

```python
print(df['match_type'].unique())
```

```python
df['result_margin'] = df['result_margin'].astype(int)
df['target_runs']   = df['target_runs'].astype(int)
df['target_overs']  = df['target_overs'].astype(int)
```

```python
print(df.dtypes)
```

```python
df['super_over'] = df['super_over'].map({'Y': 1, 'N': 0})
```

```python
#new column
df['toss_winner_won'] = (df['toss_winner'] == df['winner']).astype(int)
```

```python
#re naming columns
df['home_team'] = df['team1']
df['away_team'] = df['team2']
```

```python
# new columns
df['win_by_runs']    = (df['result'] == 'runs').astype(int)
df['win_by_wickets'] = (df['result'] == 'wickets').astype(int)
```

```python
df.head()
```

```python
print(df.dtypes)
```

```python
def team1_result(row):
    if row['winner'] == row['home_team']:
        return 'Win'
    elif row['winner'] == 'No Result':
        return 'No Result'
    else:
        return 'Loss'
df['home_team_result'] = df.apply(team1_result, axis=1)

#new column
```

```python
df['home_team'].unique()
```

```python
df = df.drop(columns=['team1', 'team2'])
```

```python
team_rename = {
    'Delhi Daredevils'           : 'Delhi Capitals',
    'Kings XI Punjab'            : 'Punjab Kings',
    'Rising Pune Supergiant'     : 'Rising Pune Supergiants',
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru',
    'Pune Warriors'              : 'Pune Warriors India',
}

cols=['toss_winner','winner','home_team','away_team']

for col in cols:
    df[col]=df[col].replace(team_rename)
```

```python
df['toss_winner'].unique()
```

```python
# Option A: CSV
df.to_csv('matches_clean.csv', index=False)
# Option B: Excel (recommended — preserves data types)
df.to_excel('matches_clean.xlsx', index=False, engine='openpyxl')

print('Export complete!')
print(f'Final shape: {df.shape}')
print(f'Columns: {list(df.columns)}')
```
